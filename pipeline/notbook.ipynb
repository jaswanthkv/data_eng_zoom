```python
prefix = "https://github.com/DataTalksClub/nyc-tlc-data/releases/download/yellow/"
url = prefix + "yellow_tripdata_2021-01.csv.gz"

df_iter = pd.read_csv(
    url,
    dtype = dtype,
    iterator=True,
    chunksize=100000
)
```

```python
!uv add tqdm
```

```python
from tqdm.auto import tqdm
```

```python
for df_chunk in tqdm(df_iter):
    df_chunk.to_sql(
        name="yellow_taxi_data",
        con=engine,
        if_exists="append"
    )
```
